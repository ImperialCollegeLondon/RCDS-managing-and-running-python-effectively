## Uses of Notebooks

Notebooks are not well-suited to a large research code, which may span dozens or hundreds of files and be many thousands of lines long. This is for a few reasons:

* It's not straightforward to run a notebook from the command line, including on remote servers, such as HPC servers.
* It's not straightforward to import code from a notebook into another notebook or into a `.py` file, making organising code into reusable modules more difficult.
* Notebooks are not well-suited to version control, as they are stored in a JSON format that is not diffable. Without going into too much detail, this means that multiple collaborators working on the same notebook can easily overwrite each other's work, and it is difficult to see what changes have been made to a notebook over time.

However, Notebooks are good for straightforward data analysis and visualisation, where there are not many lines of code and progression through the workflow is linear. They are also good for creating tutorials and teaching materials, as they allow you to combine code, text, and images in a single document. They can also be useful to present a user-friendly front-end to a more complex codebase, where the user can run code and see results without needing to understand the underlying code (which may be spread across many `.py` files).
